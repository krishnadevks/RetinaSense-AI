```python

import pandas as pd
import re

# Load dataset
df = pd.read_csv(
    '/content/drive/MyDrive/ODIR_dataset/extracted/full_df.csv'
)

# Select the correct diagnostic keyword for each image
df['eye_side'] = df['filename'].str.extract(r'_(left|right)\.jpg$')

df['eye_keyword'] = df.apply(
    lambda row: row['Left-Diagnostic Keywords']
    if row['eye_side'] == 'left'
    else row['Right-Diagnostic Keywords'],
    axis=1
)

# Clean and split diagnostic keywords
def split_keywords(value):
    if pd.isna(value):
        return []
    return [
        x.strip().lower()
        for x in re.split(r'[，,]', str(value))
        if x.strip()
    ]

df['keyword_list'] = df['eye_keyword'].apply(split_keywords)

# DR diagnostic terms
dr_terms = {
    'diabetic retinopathy',
    'mild nonproliferative retinopathy',
    'moderate non proliferative retinopathy',
    'severe nonproliferative retinopathy',
    'proliferative diabetic retinopathy',
    'severe proliferative diabetic retinopathy'
}

# Strict classification: only single-diagnosis records
def classify_eye(keywords):

    if len(keywords) != 1:
        return None

    keyword = keywords[0]

    if keyword == 'normal fundus':
        return 'Normal'

    elif keyword in dr_terms:
        return 'Diabetic Retinopathy'

    elif keyword == 'glaucoma':
        return 'Glaucoma'

    return None

df['clean_label'] = df['keyword_list'].apply(classify_eye)

# Display distribution
print("CLEAN OCULAR DATASET DISTRIBUTION")
print("================================")

print(df['clean_label'].value_counts(dropna=False))

print("\nSELECTED IMAGES:")
print(df['clean_label'].value_counts())

print("\nEXCLUDED IMAGES:")
print(df['clean_label'].isna().sum())
```

```python

import os

train_path = '/content/drive/MyDrive/ODIR_dataset/extracted/ODIR-5K/ODIR-5K/Training Images'
test_path = '/content/drive/MyDrive/ODIR_dataset/extracted/ODIR-5K/ODIR-5K/Testing Images'
preprocessed_path = '/content/drive/MyDrive/ODIR_dataset/extracted/preprocessed_images'

train_files = set(os.listdir(train_path))
test_files = set(os.listdir(test_path))
preprocessed_files = set(os.listdir(preprocessed_path))

# Identify where each image exists
def find_source(filename):
    if filename in train_files:
        return 'Training'
    elif filename in test_files:
        return 'Testing'
    elif filename in preprocessed_files:
        return 'Preprocessed only'
    return 'Missing'

df['source'] = df['filename'].apply(find_source)

# Extract patient ID from filename
df['patient_id'] = df['filename'].str.extract(r'^(\d+)_')

selected = df[df['clean_label'].notna()].copy()

print("IMAGE AVAILABILITY")
print("==================")
print(selected['source'].value_counts())

print("\nPATIENT COUNTS BY CLASS")
print("=======================")

for label, group in selected.groupby('clean_label'):
    print(
        label,
        "| Images:", len(group),
        "| Unique patients:", group['patient_id'].nunique()
    )

print("\nMISSING IMAGES:",
      (selected['source'] == 'Missing').sum())

print("\nPATIENTS WITH MULTIPLE LABELS:")
labels_per_patient = selected.groupby('patient_id')['clean_label'].nunique()

print((labels_per_patient > 1).sum())
```

```python

from sklearn.model_selection import GroupShuffleSplit

# Keep only selected ocular images
selected = df[df['clean_label'].notna()].copy()
selected = selected.reset_index(drop=True)

# Patient IDs
groups = selected['patient_id']

# First: 85% train + validation, 15% test
split1 = GroupShuffleSplit(
    n_splits=1,
    test_size=0.15,
    random_state=42
)

train_val_idx, test_idx = next(
    split1.split(selected, groups=groups)
)

train_val = selected.iloc[train_val_idx].copy()
test = selected.iloc[test_idx].copy()

# Second: 70% train, 15% validation
split2 = GroupShuffleSplit(
    n_splits=1,
    test_size=0.17647,
    random_state=42
)

train_idx, val_idx = next(
    split2.split(
        train_val,
        groups=train_val['patient_id']
    )
)

train = train_val.iloc[train_idx].copy()
val = train_val.iloc[val_idx].copy()

# Display results
for name, data in [
    ('TRAINING', train),
    ('VALIDATION', val),
    ('TESTING', test)
]:
    print('\n', name)
    print('Images:', len(data))
    print('Patients:', data['patient_id'].nunique())
    print(data['clean_label'].value_counts())

# Check patient leakage
train_patients = set(train['patient_id'])
val_patients = set(val['patient_id'])
test_patients = set(test['patient_id'])

print('\nPATIENT LEAKAGE CHECK')
print('Train-Val:', len(train_patients & val_patients))
print('Train-Test:', len(train_patients & test_patients))
print('Val-Test:', len(val_patients & test_patients))
```

```python
from tensorflow.keras.utils import load_img, img_to_array
from tensorflow.keras.applications.efficientnet import preprocess_input

def preprocess_image(image_path):
    img = load_img(image_path, target_size=(224, 224))
    img = img_to_array(img)
    img = preprocess_input(img)
    return img

sample_image = preprocess_image(train['image_path'].iloc[0])

print("Processed image shape:", sample_image.shape)
print("Data type:", sample_image.dtype)
```

```python
label_map = {
    'Normal': 0,
    'Diabetic Retinopathy': 1,
    'Glaucoma': 2
}

for dataset in [train, val, test]:
    dataset['label'] = dataset['clean_label'].map(label_map)

print(train[['clean_label', 'label']].head(10))
print("\nNumber of classes:", len(label_map))
```

```python
AUTOTUNE = tf.data.AUTOTUNE

def load_and_prepare_image(image_path, label):
    image = tf.io.read_file(image_path)
    image = tf.io.decode_jpeg(image, channels=3)
    image = tf.image.resize(image, [224, 224])
    image = tf.cast(image, tf.float32)

    return image, label


def create_dataset(dataframe, training=False):
    paths = dataframe['image_path'].values
    labels = dataframe['label'].values

    dataset = tf.data.Dataset.from_tensor_slices((paths, labels))

    dataset = dataset.map(
        load_and_prepare_image,
        num_parallel_calls=AUTOTUNE
    )

    if training:
        dataset = dataset.shuffle(buffer_size=1000)

    dataset = dataset.batch(BATCH_SIZE)
    dataset = dataset.prefetch(AUTOTUNE)

    return dataset


train_ds = create_dataset(train, training=True)
val_ds = create_dataset(val)
test_ds = create_dataset(test)

print("TensorFlow datasets created successfully!")
```

```python
from tensorflow.keras.applications import EfficientNetB0
from tensorflow.keras.layers import (
    GlobalAveragePooling2D,
    Dropout,
    Dense
)
from tensorflow.keras.models import Model

# Load pretrained EfficientNetB0
base_model = EfficientNetB0(
    weights='imagenet',
    include_top=False,
    input_shape=(224, 224, 3)
)

# Freeze pretrained layers
base_model.trainable = False

# Add our classification layers
x = base_model.output
x = GlobalAveragePooling2D()(x)
x = Dropout(0.3)(x)
output = Dense(3, activation='softmax')(x)

# Create final model
model = Model(inputs=base_model.input, outputs=output)

print("Model created successfully!")
model.summary()
```

```python
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

print("Model compiled successfully!")
```

```python
from sklearn.utils.class_weight import compute_class_weight
import numpy as np

classes = np.array([0, 1, 2])

weights = compute_class_weight(
    class_weight='balanced',
    classes=classes,
    y=train['label'].values
)

class_weights = dict(zip(classes, weights))

print("Class weights:")
for label, weight in class_weights.items():
    print(f"{label}: {weight:.4f}")
```

```python
from tensorflow.keras.callbacks import (
    EarlyStopping,
    ReduceLROnPlateau,
    ModelCheckpoint
)

callbacks = [
    EarlyStopping(
        monitor='val_loss',
        patience=3,
        restore_best_weights=True
    ),

    ReduceLROnPlateau(
        monitor='val_loss',
        factor=0.2,
        patience=2,
        min_lr=1e-6
    ),

    ModelCheckpoint(
        '/content/drive/MyDrive/ODIR_dataset/best_ocular_model.keras',
        monitor='val_loss',
        save_best_only=True
    )
]

history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=10,
    class_weight=class_weights,
    callbacks=callbacks
)
```

```python
def load_targeted_image(path, label):
    image = tf.io.read_file(path)
    image = tf.io.decode_jpeg(image, channels=3)
    image = tf.image.resize(image, [224, 224])
    image = tf.cast(image, tf.float32)

    # Apply augmentation only to DR images (label 1)
    image = tf.cond(
        tf.equal(label, 1),
        lambda: data_augmentation(
            tf.expand_dims(image, axis=0),
            training=True
        )[0],
        lambda: image
    )

    return image, label


targeted_train_ds = tf.data.Dataset.from_tensor_slices((
    train['image_path'].values,
    train['label'].values
))

targeted_train_ds = targeted_train_ds.map(
    load_targeted_image,
    num_parallel_calls=tf.data.AUTOTUNE
)

targeted_train_ds = targeted_train_ds.shuffle(1000)
targeted_train_ds = targeted_train_ds.batch(BATCH_SIZE)
targeted_train_ds = targeted_train_ds.prefetch(tf.data.AUTOTUNE)

print("Targeted DR augmentation dataset created!")
```
